```python
# Install necessary libraries
!pip install torch torchvision matplotlib pillow scikit-learn

# Imports
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import models, transforms
from torch.utils.data import DataLoader, Dataset, random_split
from PIL import Image
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
import os

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Verify images are uploaded properly
print("Uploaded images:", os.listdir("."))
# Define dataset paths and labels
from google.colab import files
uploaded = files.upload()
image_classes = {
    "normal": ["/content/drive/MyDrive/Colab Notebooks/image list/normal1.jpg.JPG"],
    "eyes_closed": ["/content/drive/MyDrive/Colab Notebooks/image list/eyes_closed1.jpg.JPG"],
    "yawning": ["/content/drive/MyDrive/Colab Notebooks/image list/yawning1.jpg.JPG"],
    "looking_down": ["/content/drive/MyDrive/Colab Notebooks/image list/looking_down1.jpg.JPG"]
}

# Custom dataset with correct loading and verification
class DrowsinessDataset(Dataset):
    def __init__(self, transform=None):
        self.transform = transform
        self.images = []
        self.labels = []
        classes = ["normal", "eyes_closed", "yawning", "looking_down"]
        for idx, cls in enumerate(classes):
            for img_file in os.listdir("."):
                if img_file.lower().startswith(cls):
                    self.images.append(img_file)
                    self.labels.append(idx)
        assert len(self.images) > 0, "No images found. Check filenames!"

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img = Image.open(self.images[idx]).convert("RGB")
        if self.transform:
            img = self.transform(img)
        label = self.labels[idx]
        return img, label

# Data transforms
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

# Initialize dataset
dataset = DrowsinessDataset(transform=transform)

# Dataset split
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = random_split(dataset, [train_size, test_size])

train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=4, shuffle=False)

# Models
teacher_model = models.resnet50(pretrained=True)
teacher_model.fc = nn.Linear(2048, 4)
teacher_model.to(device).eval()

student_model = models.mobilenet_v3_large(pretrained=True)
student_model.classifier[3] = nn.Linear(student_model.classifier[3].in_features, 4)
student_model.to(device)

# Optimizer and loss
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(student_model.parameters(), lr=0.001)

# Knowledge Distillation loss
def distillation_loss(student_logits, teacher_logits, T=3.0):
    student_probs = F.log_softmax(student_logits/T, dim=1)
    teacher_probs = F.softmax(teacher_logits/T, dim=1)
    return F.kl_div(student_probs, teacher_probs, reduction='batchmean') * (T**2)

# Training loop with metrics
epochs = 10
train_losses, test_losses, accuracies, precisions, recalls, f1_scores = [], [], [], [], [], []

for epoch in range(epochs):
    student_model.train()
    epoch_loss = 0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        with torch.no_grad():
            teacher_logits = teacher_model(images)

        student_logits = student_model(images)
        ce_loss = criterion(student_logits, labels)
        kl_loss = distillation_loss(student_logits, teacher_logits)
        loss = 0.7 * ce_loss + 0.3 * kl_loss

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_train_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_train_loss)

    # Evaluate model
    student_model.eval()
    preds, truths = [], []
    test_loss = 0

    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = student_model(images)
            loss = criterion(outputs, labels)
            test_loss += loss.item()
            preds += outputs.argmax(dim=1).cpu().tolist()
            truths += labels.cpu().tolist()

    avg_test_loss = test_loss / len(test_loader)
    test_losses.append(avg_test_loss)

    acc = accuracy_score(truths, preds)
    precision, recall, f1, _ = precision_recall_fscore_support(
        truths, preds, average='weighted', zero_division=0
    )

    accuracies.append(acc)
    precisions.append(precision)
    recalls.append(recall)
    f1_scores.append(f1)

    print(f"Epoch {epoch+1}/{epochs}")
    print(f"Train Loss: {avg_train_loss:.4f}, Test Loss: {avg_test_loss:.4f}")
    print(f"Accuracy: {acc:.4f}, Precision: {precision:.4f}, Recall: {recall:.4f}, F1-score: {f1:.4f}\n")

# Plot Loss and Metrics
plt.figure(figsize=(12, 8))
plt.subplot(2, 1, 1)
plt.plot(train_losses, label='Train Loss', marker='o')
plt.plot(test_losses, label='Test Loss', marker='s')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss Curve')
plt.legend()
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(accuracies, label='Accuracy', marker='o')
plt.plot(precisions, label='Precision', marker='s')
plt.plot(recalls, label='Recall', marker='^')
plt.plot(f1_scores, label='F1-score', marker='d')
plt.xlabel('Epoch')
plt.ylabel('Metrics')
plt.title('Performance Metrics')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()

# Save the trained student model
torch.save(student_model.state_dict(), 'student_drowsiness_model.pth')
```
